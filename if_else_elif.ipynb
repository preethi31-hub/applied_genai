# ***if elif else ladder***

```python
# creating grading system
# > 90 'A', 75 to 90 'B', 60 to 75'C', 35 to 60 'D', 20 to 35 'E', 0 to 20 'F'

score = 74
if (score >=90) and (score <=100): #upper bound
  print('A')
elif (score >=75) and (score < 90):
  print('B')
elif (score >=60) and (score < 75):
  print('C')
elif (score >=35) and (score < 60):
  print('D')
elif (score >=20) and (score < 35):
  print('E')
elif (score <20) and (score >=0):
  print('F')
else:
  print('Invalid input')
```

# ***Nested condition***

```python
age=26
indian_citizen = False
if age >=18 and indian_citizen:
  print("person is indian citizen and >= 18 years, so eligible to vote")
else:
  print("not eligible to vote")
```

```python
age=10
indian_citizen = True
if age >=18:
  print('person is having age > 18. need to check the citizenship')
  if indian_citizen:
    print("person is indian citizen and >= 18 years, so eligible to vote")
  else:
    print("person > 18, but he is not indian citizen")
else:
  print("not eligible to vote")
```
