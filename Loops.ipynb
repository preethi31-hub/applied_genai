# ***While Loop***

```python
# print values from 1 to 20
# initialization, Condition, increment/decrement (has to be there to execute while loop)
c=1
while c<=20:
  print(c)
  c=c+1
```

```python
# print value from 20 to 1
a=20
while a>=1:
  print(a)
  #a=a-1
  a-=1
```

# ***For Loop***

***For Loop with List***

```python
fruits = ["apple", "banana", "cherry"]
```

```python
fruits[1]
```

```python
for i in fruits:
  print(i)
```

***For Loop with range***

```python
# print values from 1 to 10
for i in range(1,11,1):
  print(i)
```

```python
# print all the even numbers between 1 to 20 using only range function
for a in range(2,21,2):
  print(a)
```

```python
# print from 1 to 10
for i in range(11):
  print(i)
```

```python
# print 0 t0 5 using range
for i in range(6):
  print(i)
```

```python
# print numbers from 22 to 11 in reverse order
for i in range(22,10,-1):
  print(i)
```

```python
if (10%2==0):
  print("even")
else:
  print("odd")
```

```python
# 1. print from 1 to 10 but i dont need 5
for i in range(1,11):
  if i !=5:
    print(i)

```

```python
# 2.
for i in range(1,11):
  if i==5:
    pass # nothing will print
  else:
    print(i)
```

```python
for i in range(1,11):
  if i==5:
    continue # skip 5
  else:
    print(i)
```

```python
# print 5,15,20,25,30
for i in range (5,35,5):
  if i==10:
    continue
  else:
    print(i)
```

```python
for i in range(1,11):
  if i==5:
    break #stop
  else:
    print(i)
```

```python
# run a loop from 1 to 100 but it need to stop at 20
for a in range(1,101):
  if a==21:
    break
  print(a)

```
