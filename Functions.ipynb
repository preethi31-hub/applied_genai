# ***Playing with Parameters***

```python
# addition of two numbers

def add_two_num(val1,val2):
  a1 = val1
  print("input 1:", a1)
  a2 = val2
  print("input 2:", a2)
  a3=a1+a2
  print ("addition of two numbers :",a3)
```

```python
add_two_num(10,20) # positional argument
```

```python
add_two_num(20,10)
```

```python
add_two_num(val2=20,val1=10)
```

```python
def add_two_num(val1,val2=10): #default parameter
  a1 = val1
  print("input 1:", a1)
  a2 = val2
  print("input 2:", a2)
  a3=a1+a2
  print ("addition of two numbers :",a3)
```

```python
add_two_num(20)
```

```python
add_two_num(20,40) # 40 overrides default parameter value
```

```python
def add_two_num(val1=10,val2):
  a1 = val1
  print("input 1:", a1)
  a2 = val2
  print("input 2:", a2)
  a3=a1+a2
  print ("addition of two numbers :",a3)
```

```python
add_two_num(20)
```

# ***Return Keyword***

```python
# multiplication of two numbers

def mul_two_num(num1,num2):
  x1 = num1
  print("input 1:", x1)
  x2 = num2
  print("input 2:", x2)
  x3=x1*x2
  print ("Multiplication of two numbers :",x3)
  return x3
```

```python
m = mul_two_num(10,20)
```

```python
print(m)
```

```python
# Assessment 1

# write a function that returns the total cost of the purchase. parameter required are price and quantity

def purchase_total(price,quantity):
  a1=price
  print("price :", a1)
  a2=quantity
  print("quantity :", a2)
  a3=a1*a2
  print("the total cost of the purchase :", a3)
  return a3
```

```python
m = purchase_total(100,2)
```
